### Q1. NLP Preprocessing Pipeline

Write a Python function that performs basic NLP preprocessing on a sentence. The function should do the following:

1. **Tokenize** the sentence into individual words and punctuation.
2. **Remove common English stopwords** such as `the`, `in`, and `are`.
3. **Apply stemming** to reduce each remaining word to its stem.

**Use the sentence:**

> NLP techniques are used in virtual assistants like Alexa and Siri.

**The function should print:**
1. A list of all tokens.
2. The list after stop words are removed.
3. The final list after stemming.

**Expected Output**

Your program should print three lists in this order:

1. **Original Tokens** — All words and punctuation split from the sentence.
2. **Tokens Without Stopwords** — The words remaining after removing stopwords and punctuation.
3. **Stemmed Words** — The resulting stems of the remaining words.


````python
import nltk
````

````python
from nltk.corpus import stopwords
````

````python
from nltk.stem import PorterStemmer
````

````python
from nltk.tokenize import wordpunct_tokenize
````

````python
nltk.download("stopwords",quiet=True)
````

````python
def preprocess_sentence(sentence):
    # 1. Split the sentence into words and punctuation.
    tokens = wordpunct_tokenize(sentence)

    # 2. Remove stopwords and punctuation.
    english_stopwords = set(stopwords.words("english"))
    filtered_tokens = [
        token for token in tokens
        if token.isalpha() and token.lower() not in english_stopwords
    ]

    # 3. Apply stemming to the remaining words.
    stemmer = PorterStemmer()
    stemmed_words = [stemmer.stem(token) for token in filtered_tokens]

    print("Original Tokens:", tokens)
    print("Tokens Without Stopwords:", filtered_tokens)
    print("Stemmed Words:", stemmed_words)


sentence = "NLP techniques are used in virtual assistants like Alexa and Siri."
preprocess_sentence(sentence)
````

### Q2. Scaled Dot-Product Attention

Implement scaled dot-product attention in Python using the query matrix 'Q',
the key matrix 'K', and the value matrix 'V'.

1. Compute the dot-product scores \(Q and K^T).
2. Scale the result by dividing it by (\sqrt{d}\), (where d is the key dimension).
3. Apply softmax to get attention weights.
4. Multiply the weights by V to get the output.

**Use the following test inputs**

```python
Q = np.array([[1, 0, 1, 0], [0, 1, 0, 1]])
K = np.array([[1, 0, 1, 0], [0, 1, 0, 1]])
V = np.array([[1, 2, 3, 4], [5, 6, 7, 8]])
```

**Expected Output Description**

Your output should display:
1. The attention weights matrix(after softmax).
2. The final output matrix.

````python
import numpy as np
````

````python
from nltk.metrics.aline import V
def scaled_dot_product_attention(Q,K,V):
  # Compute the dot-product scores and scale them by sqrt(d).
  d = K.shape[1]
  scaled_scores = (Q @ K.T)/np.sqrt(d)

 # Subtract each row's maximum for numerical stability.
  shifted_scores = scaled_scores - scaled_scores.max(axis=1, keepdims=True)

  #Apply softmax seperately to each query's score.
  exp_scores = np.exp(shifted_scores)
  weights =exp_scores/ exp_scores.sum(axis=1, keepdims= True)

  #Use the weights to combine the rows of V.
  output = weights @ V
  return weights, output


````

````python
Q = np.array([[1, 0, 1, 0], [0, 1, 0, 1]])
K = np.array([[1, 0, 1, 0], [0, 1, 0, 1]])
V = np.array([[1, 2, 3, 4], [5, 6, 7, 8]])
````

````python
weights, output = scaled_dot_product_attention(Q, K, V)
````

````python
print("Attention Weights Matrix:")
print(np.round(weights, 4))

print("\nFinal Output Matrix:")
print(np.round(output, 4))
````
